## Introduksjon til funksjoner og plotting.
### Først: Hvordan tegner vi grafer i Python?
 - Det er ulike bibliotek i Python som lar deg tegne grafer. Den vanligste pakken kalles matplotlib. Du må importere den for å ta den i bruk, dette gjør du øverst i programmet ditt ved å skrive *import matplotlib.pyplot as plt*.
 - For å bruke verktøy fra pakken, kan du nå skrive *plt.<navn på det du skal bruke>*.



```python
#Importerer matplotlib (dette er en kode-celle):
import matplotlib.pyplot as plt
```

### Det neste vi trenger nå er en funksjon vi kan tegne grafen til!
 - I Python bruker vi kommandoen *def <navn(variabel)>* når vi skal definere en funksjon. Deretter velger vi hva funksjonen skal returnere.
 - Hvis vi for eksempel skal lage en lineær-funksjon $f(x) = x + 2$, gjør vi det på følgende måte:

```python
#Definerer lineær funksjon:
def f(x):
    return x + 2
```

### Nå er vi snart klare til å tegne grafen, det gjenstår bare en ting!
 - Den store forskjellen på plotting i Python og for eksempel GeoGebra, er at vi i Python må lage en liste med alle x-verdier vi vil tegne grafen for, og en liste med alle tilhørende y-verdier.
 - Det er forskjellige måter å gjøre dette på, her skal vi bruke **tomme lister**, en **while-løkke**, og kommandoen **append**, som legger til verdier bakers i en liste.

```python
#Vi begynner med to helt tomme lister:
xverdier = []
yverdier = []

#Så bestemmer vi hvilken x vi vil starte på, hvilken x vi vil slutte på, og hvor store steg det skal være mellom hvert punkt:
xstart = 0
xslutt = 10
xsteg = 0.01

#Definerer første x, og første y:
x = xstart
y = f(x) #bruker funksjonen vår

#Nå fyller vi resten med en while-løkke og append():
while x < xslutt:
    xverdier.append(x) #legger x inn i xverdier
    yverdier.append(y) #legger y inn i yverdier
    x = x + xsteg #flytter x til neste steg
    y = f(x) #finner ny y-verdi
```

### Nå er listene våre lagd, der vi har x-verdier og y-verdier. Da kan vi endelig tegne grafen!
 - Nå er det tid for å bruke matplotlib-bilbioteket, som vi henter ved å skrive *plt*.
 - Kommandoen *plt.plot(x-verdier,y-verdier)* tegner grafen for de gitte x- og y-verdiene (x først så y).
 - Kommandoen *plt.legend()* viser frem merkelappen (label) som vi gir grafen.
 - Kommandoen *plt.grid()* tegner rutenett.
 - Kommandoen *plt.show()* viser figuren.
 - Det er et hav av andre måter du kan få figuren til å se finere ut på, du finner masse ved å søke litt på nettet.

```python
plt.plot(xverdier, yverdier, label = 'f(x) = x + 2') #Tegner grafen f(x)
plt.legend() #Viser merkelappen.
plt.grid() #Rutenett.
plt.show() #Viser figuren.
```
