## Numerisk integrasjon:
Vi bruker definisjonen av den deriverte, 

$f'(x)=\lim_{\Delta x\rightarrow 0}\frac{f(x+\Delta x)-f(x)}{\Delta x}$

til å regne ut en tilnærmingsverdi av den deriverte i Python. 

Dette kalles **numerisk derivasjon**. 

Når vi deriverer med derivasjonsreglene, kalles det **analytisk derivasjon**.

```python
import math
import matplotlib.pyplot as plt

def f(x):
    return math.e**(2*x) + x**2

def df_analytisk(x):
    return 2*math.e**(2*x) + 2*x

def df_numerisk(f, x, delta_x):
    return (f(x+delta_x)-f(x)) / delta_x

x = 2
delta_x = 0.1

#Her bruker jeg en funksjonalitet i print() som lar meg kombinere tekst og variabler.
#Der det står {variabel}, vil verdien av variabelen skrives ut.
#Der det står {variabel:.4f} vil det skrives ut et desimaltall med 4 tall etter komma.
print(f'Analytisk, f\'({x}) = {df_analytisk(x):.4f}')
print(f'Numerisk, f\'({x}) = {df_numerisk(f, x, delta_x):.4f}')
```

**Vi kan tegne grafen med samme metode som vi bruke sist programmeringsøkt: lage og fylle lister.**

```python
#Vi begynner med to helt tomme lister:
xverdier = []
yverdier = []

xstart = 0
xslutt = 3
xsteg = 0.01
delta_x = 0.0001

#Definerer første x, og første y:
x = xstart
y = df_numerisk(f, x, delta_x) #bruker den numeriske derivasjons-funksjonen vår

#Nå fyller vi resten med en while-løkke og append():
while x < xslutt:
    xverdier.append(x) #legger x inn i xverdier
    yverdier.append(y) #legger y inn i yverdier
    x = x + xsteg #flytter x til neste steg
    y = df_numerisk(f, x, delta_x) #finner ny y-verdi

plt.plot(xverdier, yverdier, label = '$f\'(x)$') #Tegner grafen f(x)
plt.legend() #Viser merkelappen.
plt.grid() #Rutenett.
plt.title('Numerisk derivasjon av f(x)')
plt.show() #Viser figuren.
```

## Derivere verditabeller direkte
Hvis vi har funksjonen vår som en verditabell med x-verdier og tilhørende y-verdier, kan vi estimere den deriverte direkte som $\frac{\Delta y}{\Delta x}$. 

Dette kan vi også bruker når vi deriverer ekte, ikke-analytiske datasett!

## Oppgave
I koden under er funksjonen $f(x) = x^2+2x$ representert som verditabell. Forklar hva som skjer i hver linje.

```python
xverdier = [-1, 0, 1, 3, 3.5, 4, 5, 5.1, 5.5]
yverdier = [-3, -1, 3, 17, 21.75, 27, 39, 40.31, 45.75]

print('x    f\'(x)')
print('----------')

#len(liste) gir lengden på listen. Hvorfor bruker vi dette her?
for i in range(len(xverdier) - 1): #Hvorfor -1?
    x1 = xverdier[i] #liste[tall] lar oss velge et element i listen, liste[0] gir første element.
    x2 = xverdier[i+1]
    y1 = yverdier[i]
    y2 = yverdier[i+1]
    delta_x = x2 - x1
    delta_y = y2 - y1
    derivert = delta_y / delta_x
    print(f'{x1}    {derivert: .2f}')
```
